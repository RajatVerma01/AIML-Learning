## Logistic Regression By Perceptron Trick
Perceptron Trick

The goal is to find a decision boundary that separates the two classes.

Ax+By+C=0

Initialize the parameters A, B, and C with random values.
Randomly select a training data point (x,y).
Use the current decision boundary to predict its class.
Compare the predicted class with the actual class.
If the prediction is correct, no changes are made to the parameters.
If the prediction is incorrect, update the parameters so that the decision boundary moves toward the correct classification.
Repeat this process for many training examples until all points are classified correctly (for linearly separable data) or until a stopping criterion is met.


```python
from sklearn.datasets import make_classification
import numpy as np
import matplotlib.pyplot as plt
X, y = make_classification(
    n_samples=500,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_clusters_per_class=1,
    class_sep=1.5,
    random_state=42
)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def perceptron(X,y):
    
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1
    
    for i in range(1000):
        j = np.random.randint(0,500)
        y_hat = step(np.dot(X[j],weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]
        
    return weights[0],weights[1:]
        
```

```python
def step(z):
    return 1 if z>0 else 0
```

```python
intercept,coef=perceptron(X,y)
```

```python
m=-(coef[0]/coef[1])
b=-(intercept/coef[1])
```

```python
x_input = np.linspace(-3,3,500)
y_input = m*x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=500)
plt.ylim(-3,2)
```
